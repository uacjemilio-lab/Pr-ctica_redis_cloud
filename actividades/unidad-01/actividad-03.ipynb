```python
# Importa las bibliotecas necesarias
import sqlite3
import pandas as pd

# Construye la ruta a la base de datos
ruta = "/content/ingenieria-de-datos/datos/entrada/chinook.sqlite"

# Establece la conexión
conexion = sqlite3.connect(ruta)

# Consulta las tablas disponibles
consulta = "select * from sqlite_master where type='table';"
tablas = pd.read_sql_query(consulta, conexion) # Esta linea sirve para mandar nuestra consulta a la conexion, y el resultado es almacenado en la variable tablas, que es un dataframe.

# Muestra las tablas
tablas

# Muestra las tablas disponibles
```

**Importante:** No utilices JSONPlaceholder, Open-Meteo Weather API ni NASA Open APIs, ya que fueron utilizadas en el laboratorio.

```python
# Importa las bibliotecas necesarias

import requests
# Nuestro End Point
url = "http://universities.hipolabs.com/search"
# Los parámetros de nuestra consulta
params = {"country": "Mexico"}


# Envíamos nuestra consulta al End Point, junto con nuestros parámetros.
response = requests.get(url, params=params)
# La respuesta se guarda en la var response

# Condicionales
if response.status_code == 200: # El código 200 es un código que indica éxito en la operación
    universidades = response.json()
    for uni in universidades[:5]: # Mostrar las primeras 5
        print(f"Nombre: {uni['name']}")
        print(f"Dominio: {uni['domains']}")
        print("---")
else:
    print(f"Error: {response.status_code}") # En caso de que nuestra consulta no sea éxitosa, debemos conocer el código de error.
```
